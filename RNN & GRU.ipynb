```python
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Input,GRU
import numpy as np
```

```python
from keras.datasets import imdb
(train_data, train_labels), (test_data, test_labels) = imdb.load_data(num_words=25000)
```

```python
from keras.utils import pad_sequences
x_train = pad_sequences(train_data,padding='post',maxlen=100)
x_test=pad_sequences(test_data,padding='post',maxlen=100)
```

```python
np.max(x_train)
x_train.shape
```

```python
model=Sequential()
model.add(Input(shape=(100,)))
model.add(Embedding(25000,output_dim=2))
model.add(SimpleRNN(20,return_sequences=False))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
history=model.fit(x_train,train_labels,epochs=5,validation_data=(x_test,test_labels))
```

```python
from tensorflow.keras.callbacks import EarlyStopping
call_back=EarlyStopping(
    monitor="val_loss",
    min_delta=1,
    patience=2,
    verbose=0,
    mode="auto",
    baseline=None,
    restore_best_weights=True,
    start_from_epoch=0,
)
```

```python
import tensorflow
from tensorflow.keras.layers import Dropout
model2=Sequential()
model2.add(Input(shape=(100,)))
model2.add(Embedding(25000,output_dim=2))
model2.add(GRU(9,return_sequences=False,kernel_regularizer=tensorflow.keras.regularizers.l2(0.03)))
model2.add(Dropout(0.2))
model2.add(Dense(1,activation='sigmoid'))
```

```python
model2.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
history=model2.fit(x_train,train_labels,epochs=7,validation_data=(x_test,test_labels),callbacks=call_back)
```
