```python
essay="""e live in an era of information overload.Every second, thousands of data points are shared.We are constantly bombarded by news and opinions.
Social media feeds never stop scrolling.
In this environment, information is everywhere.
However, wisdom is becoming increasingly rare.
This is why critical thinking is essential.
It is the most important skill for the modern world.
Critical thinking is not about being negative.
It is the process of objective analysis.
It involves questioning what we see and hear.
We must look beyond catchy headlines.
We must identify the source of the data.
Is the author trying to inform us?
Or are they trying to manipulate our emotions?
Without these questions, we become passive consumers.
We lose the ability to decide for ourselves.
One major challenge is the "echo chamber."
Algorithms show us what we already like.
They reinforce our existing biases.
We rarely see the other side of an argument.
This creates a narrow view of the world.
A critical thinker actively seeks different perspectives.
They understand that truth is often complex.
By reading diverse views, we grow more empathetic.
We learn to spot logical fallacies in our own logic.
Furthermore, critical thinking protects our democracy.
Society depends on an informed citizenry.
If we cannot tell fact from fiction, we cannot lead.
Misinformation spreads faster than the truth today.
It can influence elections and public health.
It can cause unnecessary panic or division.
A person who thinks critically acts as a filter.
They stop the spread of false claims.
They demand evidence before they believe a story.
In the workplace, this skill is equally vital.
Technology is changing every single industry.
Specific facts can become outdated very quickly.
However, the ability to solve problems is permanent.
Employers value people who can analyze trends.
They want workers who can find creative solutions.
A critical thinker does not just follow orders.
They look for ways to improve the process.
In conclusion, we cannot stop the flow of data.
The digital world will only get faster.
But we can control how we process that data.
Critical thinking is our greatest defense.
It allows us to remain independent and clear-headed.
It turns a flood of noise into a path of knowledge.
We must practice this skill every single day.
Our future depends on our ability to think for ourselves.
"""
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenzier=Tokenizer()
tokenzier.fit_on_texts([essay])
```

```python
tokenzier.word_index
```

```python
sequence=[]
output=[]
for sentence in essay.split('\n'):
   tokenzied=tokenzier.texts_to_sequences([sentence])[0]
   for i in range(1,len(tokenzied)):
     sequence.append(tokenzied[:i+1])

```

```python
max_len=max([len(x) for x in sequence])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
pad_arr=pad_sequences(sequence,maxlen=max_len,padding='pre')
```

```python
X=pad_arr[:,:-1]
Y=pad_arr[:,-1]
```

```python
from tensorflow.keras.utils import to_categorical
y =to_categorical(Y,num_classes=len(tokenzier.word_index)+1)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
model=Sequential()
model.add(Embedding(232,100,input_length=23))
model.add(LSTM(120))
model.add(Dense(232,activation='softmax'))
model.build(input_shape=(None, 23)) # Explicitly build the model
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=100)
```

```python
import numpy as np
text="media"
token_text=tokenzier.texts_to_sequences([text])[0]
pad_text=pad_sequences([token_text],maxlen=23,padding='pre')
pos=np.argmax(model.predict(pad_text))
for word,index in tokenzier.word_index.items():
  if index ==pos:
    print(word)
```
